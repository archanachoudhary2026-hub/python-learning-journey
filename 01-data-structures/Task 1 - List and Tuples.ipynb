# Python collections

Practice with lists, tuples, sets and dictionaries. A small product catalogue combines product names, prices and categories.

Demo data. Review copy with original assignment wording removed.


```python
products = ["Samsung Galaxy S26 Ultra", "Shampoo", "Laptop", "Tooth Brush", "Coca Cola","Soaps"]
print(products)
```

```python
sample_product = ("Shampoo",999,"Personal care")
```

```python
print("The second product in list is :",products[1])
print("The last product in the list is :", products[-1])
```

```python
products.append("Mouse")
products.append("Books")
print(products)
```

```python
new_sample_product = list(sample_product) 

new_sample_product[1] = 1999 
print(new_sample_product)
```

```python
sample_product = tuple(new_sample_product) 
print(sample_product)
```
