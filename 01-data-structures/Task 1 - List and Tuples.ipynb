# Task 1  Product Collections List and Tuples

### 1. Create a list named products containing at least 6 product names (strings).

```python
products = ["Samsung Galaxy S26 Ultra", "Shampoo", "Laptop", "Tooth Brush", "Coca Cola","Soaps"]
print(products)
```

### 2. Create a tuple named sample_product that stores (product_name, price, category) for one product.

```python
sample_product = ("Shampoo",999,"Personal care")
```

#### 3. Print the 2nd and last product from the products list

```python
print("The second product in list is :",products[1])
print("The last product in the list is :", products[-1])
```

#### 4. Append two new product names to products and then print the updated list.

```python
products.append("Mouse")
products.append("Books")
print(products)
```

#### 5. Extra (optional): Convert sample_product into a list, change its price, and convert it back to a tuple.

```python
new_sample_product = list(sample_product) 

new_sample_product[1] = 1999 
print(new_sample_product)
```

```python
sample_product = tuple(new_sample_product) 
print(sample_product)
```
