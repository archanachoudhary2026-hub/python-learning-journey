### From your products list, create a set of categories called categories_set. (If product names do not contain categories, create a short parallel list categories = [..] with matching length and use that.)

```python


categories = ["phones","Personal care","Electronics","Personal care","food & beverages","Personal care"]
categories_set = set(categories)
print(categories_set)

```

### Demonstrate adding a new category to the set and show that duplicates are ignored.

```python
categories_set.add("Footwear") 
categories_set.add("stationary")
categories_set.add("phones") 
print(f"After adding the new catorgories {categories_set}")
```

### Show how to check whether a category exists in the set (print a boolean result)

```python
categories_set = {'stationary', 'Footwear', 'food & beverages', 'phones', 'Personal care', 'Electronics'}

        
exists_check = "phones" in categories_set
print(exists_check)

exists_check = "Clothing" in categories_set
print(exists_check)
```
