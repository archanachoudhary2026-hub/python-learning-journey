### 3. Print the average price of all products (use only dictionary operations and basic arithmetic)

```python

price_dict = {'Samsung Galaxy S26 Ultra': 90000, 'Shampoo': 1999, 'Laptop': 55000, 'Soaps': 150, 'comb': 250}
total_of_price = 0

for value in price_dict.values():
   total_of_price = total_of_price + value 
Average_price  = total_of_price / len(price_dict)
print(Average_price)
```

## Extra (optional): Print the product with both the maximum and minimum prices.

```python
price_dict = {'Samsung Galaxy S26 Ultra': 90000, 'Shampoo': 1999, 'Laptop': 55000, 'Soaps': 150, 'comb': 250}



max_product_name  = max(price_dict, key = price_dict.get)
min_product_name = min(price_dict, key = price_dict.get)



max_price = price_dict[max_product_name]
min_price = price_dict[min_product_name]



print(f"{max_product_name} has the highest price {max_price } and {min_product_name} has the lowest price {min_price }")
    
```
