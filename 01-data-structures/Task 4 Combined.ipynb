# Python collections

Practice with lists, tuples, sets and dictionaries. A small product catalogue combines product names, prices and categories.

Demo data. Review copy with original assignment wording removed.


```python
products =['Samsung Galaxy S26 Ultra', 'Shampoo', 'Laptop', 'Tooth Brush', 'Coca Cola', 'Soaps', 'Mouse', 'Book']
categories = ["phones","Personal care","Electronics","Personal care","food & beverages","Personal care", "Electronics","Stationary"]

price_dict ={"Samsung Galaxy S26 Ultra": 90000, "Shampoo":500, "Laptop":55000, "Tooth Brush":150, "Coca Cola":20,"Soaps":150,"Mouse":750,"Book":1000}
price_list = list(price_dict.values())

catalog = []

for items in range(len(products)):
    product = products[items]
    category = categories[items]
    price = price_dict[product]
    catalog.append((product,price,category))
    
print(catalog)  
```

```python
category_to_products = {}

for product, price, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = [] 
    category_to_products[category].append(product) 
     
print(category_to_products)


```

```python
max_count = -1
largest_category =""

for keys, values in category_to_products.items():
    if len(values) > max_count:
        max_count =len(values)
        largest_category = keys
        
print(f"{largest_category} is the largest catergory with highest number of products and  \n lists of products in that category is {category_to_products[largest_category]}")
```
