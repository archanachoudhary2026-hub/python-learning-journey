# Task 1: Discount Rules (if / elif / else)


##### Write a program that reads an integer order_amount from the user using input().
##### Apply the following discount rules and print the final amount:
##### order_amount >= 2000 → 15% 
##### discount 1500 <= order_amount < 2000 → 10% 
##### discount 1000 <= order_amount < 1500 → 7% discount
##### otherwise → 0% discount

##### Ensure you convert input to a number and handle the case when the input is non-numeric by displaying an error message and exiting the program.
##### Extra (optional): Add tax (fixed 5%) after discount and print subtotal, tax, and final total (still within conditionals and basic arithmetic).

```python
try: 
    Order_amount = int(input("Please enter the order amount"))
    if Order_amount >= 2000:
        final_amount= Order_amount -(Order_amount*0.15)
        print(f"Order value = {Order_amount}, Discount = 15% and Final amount is {final_amount}")
    elif Order_amount >=1500:
        final_amount= Order_amount -(Order_amount*0.10)
        print(f"Order value = {Order_amount}, Discount = 10% , and Final amount is {final_amount}")
    elif Order_amount >=1000:
         final_amount= Order_amount -(Order_amount*0.07)
         print(f"Order value = {Order_amount}, Discount = 7% , and Final amount is {final_amount}")
    else:
        final_amount= Order_amount -(Order_amount*0.00)
        print(f"Order value = {Order_amount}, Discount = 0% ,  and Final amount is {final_amount}")
except ValueError:
    print("Please enter a valid input")
                       
```
