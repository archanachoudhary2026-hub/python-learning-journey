# Conditions and loops

Practice with discount rules, repeated calculations, menu choices and stopping or skipping loop entries.

Demo data. Review copy with original assignment wording removed.


```python

order_amount = [1200, 2500, 800, 1750, 3000]

Total_Reveune = 0

for order in order_amount:
    if order >=2000:
        final_amount = order - (order*0.15)
        print(f"Order Amount = {order}, Discount is 15% , Final amount = {final_amount}")
    elif order >=1500:
        final_amount = order -(order*0.10)
        print(f"Order Amount = {order}, Discount is 10% , Final amount = {final_amount}")
    elif order >=1000:
        final_amount = order -(order*0.07)
        print(f"Order Amount = {order}, Discount is 7% , Final amount = {final_amount}")
    else:
        final_amount = order -(order*0.0)
        print(f"Order Amount = {order}, Discount is 0% , Final amount = {final_amount}")
    Total_Reveune = Total_Reveune+final_amount
print(f"Total Revenue is {Total_Reveune}")
```
