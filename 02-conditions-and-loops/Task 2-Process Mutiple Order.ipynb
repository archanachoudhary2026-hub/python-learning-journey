## Task 2: Process Multiple Orders (for loop)Given a list of order amounts: orders = [1200, 2500, 800, 1750, 3000], 
#### use a for loop to apply the discount rules from Task 1 to each order and print a summary table showing: order_amount -> discount% -> final_amount.

#### Also compute and print the total revenue after discounts.Extra (optional): Print the number of orders that received a discount (discount > 0).

```python

order_amount = [1200, 2500, 800, 1750, 3000]

Total_Reveune = 0

for order in order_amount:
    if order >=2000:
        final_amount = order - (order*0.15)
        print(f"Order Amount = {order}, Discount is 15% , Final amount = {final_amount}")
    elif order >=1500:
        final_amount = order -(order*0.10)
        print(f"Order Amount = {order}, Discount is 10% , Final amount = {final_amount}")
    elif order >=1000:
        final_amount = order -(order*0.07)
        print(f"Order Amount = {order}, Discount is 7% , Final amount = {final_amount}")
    else:
        final_amount = order -(order*0.0)
        print(f"Order Amount = {order}, Discount is 0% , Final amount = {final_amount}")
    Total_Reveune = Total_Reveune+final_amount
print(f"Total Revenue is {Total_Reveune}")
```

## Extra (optional): Print the number of orders that received a discount (discount > 0).

```python
order_amount = [1200, 2500, 800, 1750, 3000]

total_order_with_discount = 0

for order in order_amount:
    if order  >=1000:
        total_order_with_discount= total_order_with_discount + 1
print(f"Total order with discount = {total_order_with_discount}")
```
