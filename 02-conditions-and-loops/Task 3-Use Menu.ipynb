## Task 3: User Menu (while loop + break/continue)Implement a simple while-loop driven menu that keeps asking the user to choose an action until they press q to quit.
#### Menu options:
#### 1 — Add order amount to a running list (use input())
#### 2 — Show all orders and totals after applying discounts
#### q — Quit

#### Use continue to re-show the menu after invalid input and break to exit on q.Constraints: Keep the loop logic simple and use only lists and loops (no functions required).

```python
Order = []

while True: 
    print("----Menu----")
    print("choose 1 : To add the order")
    print("choose q: Quit")
    print("choose 2: order list and total order value afte discount")
    print("========================")

    choice = input("Please choose from the menu")

    if choice.lower() == "q":
        print("thank you for your time. Have a good day")
        break
    
    if choice == "1":
        order_amount = input("Please enter the amount")
        if order_amount.isdigit():
            Order.append(int(order_amount))
            print("Order has been successfully added to the list")
        else:
            print("Please enter a valid amount")
            continue
    elif choice == "2":
            total_order = 0 
            for amount in Order: 
                if amount >=2000:
                    final_amount = amount -(amount*0.15)
                    total_order = total_order + final_amount
                elif amount >=1500:
                    final_amount = amount -(amount*0.10)
                    total_order = total_order + final_amount
                elif amount >=1000:
                    final_amount = amount -(amount*0.07)
                    total_order = total_order + final_amount
                else:
                    final_amount = amount
                    total_order = total_order + final_amount
            print(f'Order list is {Order} and total Order value is {total_order}')
    else:
        print("Please choose a valid menu option")

```
