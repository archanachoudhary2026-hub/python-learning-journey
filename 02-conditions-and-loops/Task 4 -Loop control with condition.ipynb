### Given a list of daily sales: daily = [200, 150, 0, 400, 50, -1, 300], 

#### iterate through it with a for loop and:
#### If a day's value is -1, treat it as corrupted data and break the loop (stop processing).If a day's #### If a day's value is 0, treat it as a day with no sales and continue (skip adding to revenue).
#### For valid  positive sales, add to total_sales and print the running total.
#### Print the final total after the loop  completes (or stops due to corrupted data).

```python
daily_sales = [200, 150, 0, 400, 50, -1, 300]

total_sales = 0

for sales in daily_sales:

    if sales < 0:
        print("Corrupt data found. Stop processing.")
        break

    if sales == 0:
        continue

    total_sales += sales
    print("Running total:", total_sales)

print("Total sales:", total_sales)
```
