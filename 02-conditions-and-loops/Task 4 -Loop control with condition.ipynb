# Conditions and loops

Practice with discount rules, repeated calculations, menu choices and stopping or skipping loop entries.

Demo data. Review copy with original assignment wording removed.


```python
daily_sales = [200, 150, 0, 400, 50, -1, 300]

total_sales = 0

for sales in daily_sales:

    if sales < 0:
        print("Corrupt data found. Stop processing.")
        break

    if sales == 0:
        continue

    total_sales += sales
    print("Running total:", total_sales)

print("Total sales:", total_sales)
```
