# Python functions

Practice with function arguments, recursion, lambda expressions, map and filter. The price menu calls separate functions for adding values, averages and maximums.

Demo data. Review copy with original assignment wording removed.


```python


def add_price(prices_list, price):
    """Adds a new price to the list."""
    prices_list.append(price)

def get_average_price(prices_list):
    """Returns the average price, or 0 if the list is empty."""
    if not prices_list:
        return 0
    return sum(prices_list) / len(prices_list)

def get_max_price(prices_list):
    """Returns the maximum price, or 0 if the list is empty."""
    if not prices_list:
        return 0
    return max(prices_list)





my_prices = []

while True:
    
    print("\n--- Menu ---")
    print("1 → Add price")
    print("2 → Show average price")
    print("3 → Show highest price")
    print("q → Quit")
    
    
    choice = input("Enter your choice: ").strip().lower()
    
    if choice == '1':
        try:
            new_price = float(input("Enter the price to add: "))
            add_price(my_prices, new_price)
            print(f"Price {new_price} added successfully.")
        except ValueError:
            print("Invalid input. Please enter a numerical value.")
            
    elif choice == '2':
        avg = get_average_price(my_prices)
        print(f"The average price is: {avg:.2f}")
        
    elif choice == '3':
        highest = get_max_price(my_prices)
        print(f"The highest price is: {highest:.2f}")
        
    elif choice == 'q':
        print("Goodbye!")
        break
        
    else:
        print("Invalid choice. Please select 1, 2, 3, or q.")
```
