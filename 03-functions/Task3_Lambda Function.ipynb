## **Task 3 - Lambda Function: GST Calculator**

Create a lambda function gst that returns price after adding 18% GST.

Example:
gst = lambda price: price + (0.18 * price)
print(gst(100)) # should return 118

Extra (optional): Create another lambda to compute final price after GST + discount together.

**lambda function gst that returns price after adding 18% GST.**

```python
gst_price = lambda price:price+(price*0.18)
```

```python
print(f"Base price 100 with 18% GST: {gst_price(100)}") 
```

# **Lambda function taking base price and discount percentage as arguments**

```python

gst_discount = lambda price, discount: (price - (price * discount / 100)) + ((price - (price * discount / 100)) * 0.18)


```

```python

print(f"Base price 100 with 10% discount and 18% GST: {gst_discount(100, 10)}")  
```
