# Python functions

Practice with function arguments, recursion, lambda expressions, map and filter. The price menu calls separate functions for adding values, averages and maximums.

Demo data. Review copy with original assignment wording removed.


```python
gst_price = lambda price:price+(price*0.18)
```

```python
print(f"Base price 100 with 18% GST: {gst_price(100)}") 
```

```python

gst_discount = lambda price, discount: (price - (price * discount / 100)) + ((price - (price * discount / 100)) * 0.18)


```

```python

print(f"Base price 100 with 10% discount and 18% GST: {gst_discount(100, 10)}")  
```
