# Python functions

Practice with function arguments, recursion, lambda expressions, map and filter. The price menu calls separate functions for adding values, averages and maximums.

Demo data. Review copy with original assignment wording removed.


```python
prices = [100, 250, 400, 1200, 50, 2000, 850]


expensive_products = list(filter(lambda price: price > 500, prices))


affordable_products = list(filter(lambda price: price <= 500, prices))

print("Prices greater than 500:", expensive_products)
print("Prices less than or equal to 500:", affordable_products)
```
