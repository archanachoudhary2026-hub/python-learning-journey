# Python functions

Practice with function arguments, recursion, lambda expressions, map and filter. The price menu calls separate functions for adding values, averages and maximums.

Demo data. Review copy with original assignment wording removed.


```python
def process_prices(prices):
    
    discounted_prices = list(map(lambda price: price * 0.90, prices))
    
    
    filtered_prices = list(filter(lambda price: price > 300, discounted_prices))
    
    
    return discounted_prices, filtered_prices


test_prices = [100, 500, 900, 50, 750]
discounted, filtered = process_prices(test_prices)

print("Discounted:", discounted)
print("Filtered:", filtered)
```
