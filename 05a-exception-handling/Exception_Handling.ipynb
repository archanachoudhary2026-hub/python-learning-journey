# **Task 1: Safe Division Utility**

Write a program that:

1. Takes two inputs from the user: numerator and denominator.
2. Uses try‑except to handle:
- `ValueError` (if input is not a number)
- `ZeroDivisionError` (if denominator = 0)
3. If no error occurs, print the result inside the `else` block.
4. In the `finally` block, print:

```python
try:
    Numerator = int(input("Enter the numerator only in number"))
    Denominator = int(input("Enter the denominator only in number other then zero"))
    result = Numerator/Denominator
except ValueError :
    print("Invalid Value: Enter only numeric value, not words")
except ZeroDivisionError:
    print("Invalid Value : Please do not enter Zero")
else:
    print(result)
finally:
    print("Exceutive is completed")
```

# **Task 2: Bill Calculator with Error Handling**

Given a list of product prices:  
`prices = [120, 350, 'abc', 500, -200, 800]`

Write code that:
1. Iterates through the list.
2. Tries to add only valid (positive numerical) prices to the total.
3. Handles:
   - `TypeError` if value is not a number
   - Custom exception using `raise ValueError("Negative price not allowed")`
4. Prints the running total.

**Expected behavior:**  
Skip invalid items but continue processing.

```python
prices = [120, 350, 'abc', 500, -200, 800]
total = 0

for price in prices:
    try:
                                                              
                                                                                  
        if price < 0:
            raise ValueError("Negative price not allowed")
            
        total += price
        print(f"Added {price}. Running Total: {total}")

    except TypeError:
        print(f"Skipping '{price}': Invalid type (not a number)")
    except ValueError as e:
        print(f"Skipping {price}: {e}")

print(f"\nFinal Valid Total: {total}")
```

# Task 3: Custom Exception: Age Validator

1. Write a function `check_age(age)` that:
   - Raises a custom exception `ValueError("Age must be between 1 and 120")` if age is out of range.

2. In your main code:
   - Take age input from the user.
   - Use try-except to catch and print the custom error message.

```python
def check_age(age):
    if age >=1 and age <= 120:
        print(f"{age} is a valid age")
    else:
        raise ValueError (f"{age} is an inavlid entry,Age must be between 1 and 120")

while True:
    try:
        age= int(input("Please enter the age in numbers"))
        check_age(age)
        break
    except ValueError as ve:
        print(ve)
```

# Task 5: Mini Program: Safe Shopping Cart

Create a program that:
1. Has a cart list → `cart = []`
2. Runs a loop asking user to enter prices.
3. Stops when user enters `'q'`.
4. Inside the loop:
   - Convert input to a float
   - Handle `ValueError` if user enters invalid input
   - Raise custom exception if price is negative
5. At the end, print:
   - Total items
   - Total bill

```python
import math

cart = []

while True:
    try:
        user_input = input("Enter price (or 'Q' to quit): ")
        
                                                         
        if user_input.strip().lower() == 'q':
            break
            
        price = float(user_input)

        if not math.isfinite(price) or price <= 0:
            raise ValueError(f"{price},Price must be finite and greater than zero")
            
        cart.append(price)

    except ValueError as ve:
        print(f"Invalid input: {ve}")

print("\n--- Cart Summary ---")
print(f"Cart items: {cart}")
print(f"Total items in the cart = {len(cart)}")
print(f"Total cart Value = {sum(cart):.2f}")
```
