## **Task 6: Magic Methods & Operator Overloading**

Add the following to your Product class:
1. __str__ method
Returns a readable string:
Product(name, price, category)
2. Operator Overloading (__add__)
Allow:
product1 + product2
To return the total combined price.
Test this with two product objects.

```python
class Product:
    def __init__(self,name,price,category):
        self.name=name
        self.price=price
        self.category=category

    def __add__(self, other):
        Total = self.price + other.price
        return Total

    def __str__(self):
        return f"Product({self.name},Rs{self.price},{self.category})"
```

```python
product1 = Product("Laptop",90000,"Computers")
product2 = Product("Samsung S26 Ultra",150000,"Phone")


print(product1)
print(product2)
```

```python
Total_combined_price = product1+product2
print(Total_combined_price)
```
