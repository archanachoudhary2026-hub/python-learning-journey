# Task-5 Abstraction

Object-oriented Python practice with demo products.

```python
from abc import ABC, abstractmethod

                     
class Payment(ABC):
    @abstractmethod
    def process_payment(self):
        pass

                          
class Creditcardpayment(Payment):
    def __init__(self, amount, status):
        self.amount = amount
        self.status = status

    def process_payment(self):
        if self.status == "Successful":
            print(f"Thank you! Your Credit card Payment of {self.amount} is {self.status}")
        else:
            print(f"Sorry Your credit card Payment of {self.amount} is {self.status}")

                  
class UPIPayment(Payment):
    def __init__(self, amount, status):
        self.amount = amount
        self.status = status

    def process_payment(self):
    
        if self.status == "Successful":
            print(f"Thank you! Your UPI Payment of {self.amount} is {self.status}")
        else:
            print(f"Sorry Your UPI Payment of {self.amount} is {self.status}")
```

```python
cc_payment1 = Creditcardpayment(5000, "Successful")
cc_payment1.process_payment()

                                            
cc_payment2 = Creditcardpayment(50000, "Unsuccessful")
cc_payment2.process_payment()

                                  
upi_payment1 = UPIPayment(1000, "Successful")
upi_payment1.process_payment()

                                    
upi_payment2 = UPIPayment(2500, "Unsuccessful")
upi_payment2.process_payment()
```
