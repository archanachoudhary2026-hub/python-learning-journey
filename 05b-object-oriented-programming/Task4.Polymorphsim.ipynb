#   Task 4: **Polymorphism**

Create two classes:
Laptop(Product)
Mobile(Product)
Both override:
def get_info(self):
# Print details in their own style
Write a loop that iterates over objects of Laptop and Mobile and calls get_info() on each to show polymorphism.

```python
class Product:

    def __init__(self, name, price, category):
        """Initializes a new Product instance with core traits."""
        self.name = name
        self.price = price
        self.category = category

    def get_info(self):
        """Prints out clear, cleanly formatted product details."""
        print(
            f"Hello your product {self.name}, in category {self.category} has a price {self.price}"
        )


class Laptop(Product):

    def __init__(self, name, price, category, color):
        super().__init__(name, price, category)
        self.color = color

    def get_info(self):
        print(
            f"Hello your product {self.name} in {self.color}, in category {self.category} has a price {self.price}. "
        )


class Mobile(Product):

    def __init__(self, name, price, category, size):
        super().__init__(name, price, category)
        self.size = size

    def get_info(self):
        print(
            f"Hello your product {self.name} in {self.size}, in category {self.category} has a price {self.price}. "
        )


                                       
products_list = [
    Laptop("MacBook Pro", 1999, "Electronics", "Space Gray"),
    Mobile("iPhone 15", 999, "Electronics", "6.1 inch"),
    Laptop("Dell XPS", 1399, "Electronics", "Silver"),
    Mobile("Galaxy S24", 799, "Electronics", "6.2 inch"),
]

for product in products_list:
    product.get_info()
```
