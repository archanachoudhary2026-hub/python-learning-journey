**Task 1: Creating NumPy Arrays**

1. Import NumPy as np.
2. Create the following arrays:
* A 1D array of integers from 1 to 10
* A 2D array of shape (3, 3) with values from 1 to 9
* A NumPy array from the list: [10, 20, 30, 40, 50]
3. Print:
* Shape of each array
* Data type of each array

```python
import numpy as np
```

```python
one_dimensional_array = np.arange(1,11)
print(one_dimensional_array)
```

```python
two_dimensional_array = np.arange(1,10).reshape(3,3)
print(two_dimensional_array)
```

```python
number_array = np.array([10, 20, 30, 40, 50])
print(number_array)
```

```python
print("=== 1D Array (1 to 10) ===")
print(f"Array: {one_dimensional_array}")
print("Shape:", one_dimensional_array.shape)
print("Data Type:", one_dimensional_array.dtype)
print("-" * 30)
```

```python
print("=== 2D Array (3x3 Matrix) ===")
print("Array:\n", two_dimensional_array)
print("Shape:    ", two_dimensional_array.shape)
print("Data Type:", two_dimensional_array.dtype)
print("-" * 30)
```

```python
print("=== Array from List ===")
print("Array:    ", number_array)
print("Shape:    ", number_array.shape)
print("Data Type:", number_array.dtype)
print("=" * 30)
```
