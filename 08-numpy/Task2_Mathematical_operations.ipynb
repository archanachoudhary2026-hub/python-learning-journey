**Task 2: Important Mathematical Operations**

Given two arrays:
A = np.array([10, 20, 30, 40])
B = np.array([1, 2, 3, 4])
Perform and print results for:
1. Addition (A + B)
2. Subtraction (A - B)
3. Multiplication (A * B)
4. Division (A / B)
5. Power (A ** 2)

```python
import numpy as np
```

```python
A = np.array([10, 20, 30, 40])
B = np.array([1, 2, 3, 4])

addition = np.add(A,B)
print(A)
print(B)
print(f"Addition is {addition}")
print("="*30)
```

```python
A = np.array([10, 20, 30, 40])
B = np.array([1, 2, 3, 4])

subraction = np.subtract(A,B)
print(A)
print(B)
print(f"Subraction  is {subraction}")
print("="*30)
```

```python
A = np.array([10, 20, 30, 40])
B = np.array([1, 2, 3, 4])

Multiplication = np.multiply(A, B)
print(A)
print(B)
print(f"Multiplication  is {Multiplication}")
print("="*30)
```

```python
A = np.array([10, 20, 30, 40])
B = np.array([1, 2, 3, 4])

Division = np.divide(A, B)
print(A)
print(B)
print(f"Division  is {Division}")
print("="*30)
```

```python
A = np.array([10, 20, 30, 40])

Power = np.power(A,2)
print(A)
print(f"Power  is {Power}")
print("="*30)
```
