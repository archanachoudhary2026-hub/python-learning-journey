**Task 3: Important NumPy Mathematical Formulas**

Given:
values = np.array([2, 4, 6, 8, 10])
Calculate:
1. Square root of each element
2. Exponential of each element
3. Natural logarithm of each element
4. Sum of all elements
5. Cumulative sum of elements
Use NumPy functions only.

```python
import numpy as np
```

```python
values = np.array([2, 4, 6, 8, 10])
square = np.round(np.sqrt(values),2)
print(f"Values= {values}")
print(square)
```

```python
values = np.array([2, 4, 6, 8, 10])
exponential_value = np.exp(values)
print(f"Values= {values}")
print(f"exponential_value  is : {exponential_value }")
```

```python
values = np.array([2, 4, 6, 8, 10])
log_value = np.log(values)
print(f"Values= {values}")
print(f"Log_value  is : {log_value }")
```

```python
values = np.array([2, 4, 6, 8, 10])
Sum_of_Values = np.sum(values)
print(f"Values= {values}")
print(f"Total : {Sum_of_Values}")
```

```python
values = np.array([2, 4, 6, 8, 10])
Cumulative_value = np.cumsum(values)
print(f"Values= {values}")
print(f"Total_cumulative : {Cumulative_value}")
```
