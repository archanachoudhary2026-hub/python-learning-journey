**Task 4: Aggregation Operations**

Given a 2D array:
data = np.array([
[40, 50, 60],
[70, 80, 90]
])
Find:
1. Row-wise sum
2. Column-wise sum
3. Minimum value
4. Maximum value
5. Overall mean

```python
import numpy as np
```

```python
data = np.array([
    [40, 50, 60],
    [70, 80, 90]
    ])

sum_row = np.sum(data,axis=1)
print("========Output========")
print(f"Data = {data}")
print(f"Row Wise Sum ={sum_row}")
```

```python
data = np.array([
    [40, 50, 60],
    [70, 80, 90]
    ])

sum_Coloum = np.sum(data,axis=0)
print("========Output========")
print(f"Data = {data}")
print(f"Coloumn Wise Sum ={sum_Coloum}")
```

```python
data = np.array([
    [40, 50, 60],
    [70, 80, 90]
    ])

minimum_value = data.min(axis=1)     

print("========Minimum value Row wise========")
print(f"Data = {data}")
print(f"Minimum Value ={minimum_value}")
print("="*30)

                        
                              

minimum_value = data.min(axis=0)     

print("========Minimum value Coloumn wise========")
print(f"Data = {data}")
print(f"Minimum Value ={minimum_value}")
print("="*30)

                        
                         

data = np.array([
    [40, 50, 60],
    [70, 80, 90]
    ])

minimum_value = data.min()     

print("========Minimum value Overall========")
print(f"Data = {data}")
print(f"Minimum Value ={minimum_value}")
print("="*30)
```

```python
data = np.array([
    [40, 50, 60],
    [70, 80, 90]
    ])

maximum_value = data.max(axis=1)     

print("========Maximum value Row wise========")
print(f"Data = {data}")
print(f"Maximum Value ={maximum_value}")
print("="*30)

                        


                                

data = np.array([
    [40, 50, 60],
    [70, 80, 90]
    ])

maximum_value = data.max(axis=0)     

print("========Maximum value Coloum wise========")
print(f"Data = {data}")
print(f"Maximum Value ={maximum_value}")
print("="*30)

                        
                         

data = np.array([
    [40, 50, 60],
    [70, 80, 90]
    ])

Maximum_value = data.max()     

print("========Maximum value Overall========")
print(f"Data = {data}")
print(f"Maximum Value ={Maximum_value}")
print("="*30)
```

```python
data = np.array([
    [40, 50, 60],
    [70, 80, 90]
    ])

Mean_value = data.mean()     

print("========Mean value Overall========")
print(f"Data = {data}")
print(f"Mean Value ={Mean_value}")
print("="*30)
```
