# Statistics with NumPy

Practice with NumPy arrays and demo values.

```python
import numpy as np
```

```python
marks = np.array([78, 85, 90, 66, 72, 88, 95, 60])

mean_value = np.mean(marks)
median_value = np.median(marks)
Variance = np.round(np.var(marks),2)
Standard_Deviation= np.round(np.std(marks),2)
Minimum = np.min(marks)
Maximum = np.max(marks)
Range = Maximum-Minimum

print("========Output Value=========")
print(" ")
print(f"Marks = {marks}")
print(" ")
print(f'Mean value: {mean_value}')
print(" ")
print(f"Median value: {median_value}")
print(" ")
print(f"Variance: {Variance}")
print(" ")
print(f"Standard_Deviation: {Standard_Deviation}")
print(" ")
print(f"Minimum Marks: {Minimum}")
print(" ")
print(f"Maximum Marks: {Maximum}")
print(" ")
print(f"Range: {Range}")
```
