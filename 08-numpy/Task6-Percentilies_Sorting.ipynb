# Percentiles and sorting

Practice with NumPy arrays and demo values.

```python
import numpy as np
```

```python
marks = np.array([78, 85, 90, 66, 72, 88, 95, 60])

sorted_marks = np.sort(marks)

Percentile_25th = np.percentile(marks,25)
Percentile_50th = np.percentile(marks,50)
Percentile_75th = np.percentile(marks,75)

Average = np.average(marks)

count_above_average = int(np.sum(marks > Average))
```

```python
print("=========Output===========")

print(f"Sorted Marks: {sorted_marks}")
print(" ")
print(f"Percentile_25th : {Percentile_25th}")
print(" ")
print(f"Percentile_50th : {Percentile_50th}")
print(" ")
print(f"Percentile_75th : {Percentile_75th}")
print(" ")
print(f"Average: {Average}")
print(" ")
print(f"Count above Average ={count_above_average}")
print("="*30)
```
