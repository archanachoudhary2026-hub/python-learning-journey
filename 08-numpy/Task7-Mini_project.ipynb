**Task 7: Mini Use Case: Sales Analysis**

Given daily sales data:
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])
Perform:
1. Total weekly sales
2. Average daily sales
3. Highest and lowest sales day
4. Standard deviation of sales
5. Identify days where sales were above average

Implementation note: this version reports the highest and lowest sales amounts and the number of above-average days. It does not label those days by name or index.

```python
import numpy as np
```

```python
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])

Total_Weekly_Sales = np.sum(sales)
Average_Weekly_Sales = np.round(np.average(sales),2)
Highsest_sales = np.max(sales)
Lowest_sales =np.min(sales)
standard_Deviation = np.round(np.std(sales),2)
Sales_above_Average = np.sum(sales>Average_Weekly_Sales)
```

```python
print("==========Sales Analysis==============")
print(" ")

print(f"Sales = {sales}")
print(" ")

print(f"Total Weekly Sales: {Total_Weekly_Sales}")
print(" ")

print(f"Average Weekly Sales: {Average_Weekly_Sales}")
print(" ")

print(F"Highest_sales = {Highsest_sales}")
print(" ")

print(f"Lowest_sales ={Lowest_sales}")
print(" ")

print(f"standard_Deviation ={standard_Deviation}")
print(" ")

print(f"Above Average Sales Days  : {Sales_above_Average}")
```
