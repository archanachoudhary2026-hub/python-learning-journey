# Pandas practice

Practice with Series and DataFrames, arithmetic, filtering, grouping and charts. The sales example totals daily demo revenue and compares days with the average.

Demo data. Review copy with original assignment wording removed.


```python
import pandas as pd
import matplotlib.pyplot as plt

students = {
'Name': ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'],
'Marks':[78, 85, 90, 66, 72],
'Subject': ['Math', 'Math', 'Science', 'Science', 'Math']
}

student_df = pd.DataFrame(students)
```

```python


(student_df.plot(x="Name",y="Marks",kind="bar"))

plt.tight_layout()
plt.show()
```

```python


(student_df.plot(x="Name",y="Marks",kind="line"))


plt.tight_layout()
plt.show()
```

```python


(student_df["Marks"].plot(kind="hist"))
plt.tight_layout()
plt.show()
```
