# Car sales charts with Matplotlib

Pandas grouping and Matplotlib line, scatter, bar, stacked bar, histogram and pie charts. The five analysis columns are included as a compressed text data file, described in README.md.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
import base64
import gzip
import io
from pathlib import Path

compressed = base64.b64decode(Path("sales_data.csv.gz.b64").read_text())
sf = pd.read_csv(io.BytesIO(gzip.decompress(compressed)))


sf["Date"] = pd.to_datetime(sf["Date"])                                                  
sf["Month"] = sf["Date"].dt.month_name().str[:3]                              
month_order= ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
sale_by_month = sf.groupby("Month")['Price ($)'].sum().reindex(month_order, fill_value=0)
```

```python
x = sale_by_month.index
y=  sale_by_month.values

plt.figure(figsize=(10,6))
plt.xlabel("Month")
plt.ylabel("sales")
plt.title("Sales by Month")
plt.plot(x, y, linestyle='--', color='red', marker='o')
plt.grid(True)
plt.show()
```

```python
Sales_by_annual_income = sf.groupby('Annual Income')['Price ($)'].sum()

                        
a = Sales_by_annual_income.index                         
b = Sales_by_annual_income.values                

              
plt.figure(figsize=(10,6))
plt.yscale('log')
plt.xlabel("Annual Income")
plt.ylabel("Sales")
plt.title("Sales by Annual Income")
plt.scatter(a, b, color='red', marker='o')
plt.grid(True)
plt.show()
```

```python
sale_by_month = sf.groupby("Month")['Price ($)'].sum().reindex(month_order, fill_value=0)                  

x = sale_by_month.index
y=  sale_by_month.values

                    
plt.figure(figsize=(10,6))
plt.xlabel("Month")
plt.ylabel("sales")
plt.title("Sales by Month")
plt.bar(x, y, color='pink')

plt.show()
```

```python
sale_by_region = sf.groupby('Dealer_Region')['Price ($)'].sum()

x = sale_by_region.index
y=  sale_by_region.values

plt.figure(figsize=(10,6))
plt.xlabel("Sales")
plt.ylabel("Dealer Region")
plt.title("Sales by Dealer Region")
plt.barh(x, y, color='green')
plt.show()
```

```python
sf["Year"] = sf["Date"].dt.year
sales_2022= sf[sf["Year"]==2022]
sales_2023 = sf[sf["Year"]==2023]
s1= sales_2022.groupby("Month")["Price ($)"].sum()
s2= sales_2023.groupby("Month")["Price ($)"].sum()


month_order= ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
s1.index = pd.CategoricalIndex(s1.index,categories=month_order,ordered=True)
s2.index = pd.CategoricalIndex(s2.index,categories=month_order,ordered=True)
s1 = s1.reindex(month_order, fill_value=0)
s2 = s2.reindex(month_order, fill_value=0)

                           
x = np.arange(len(month_order))
bar_width = 0.35

plt.figure(figsize=(12,6))
plt.bar(x - bar_width/2, s1.values, width=bar_width, label="2022", color="skyblue")
plt.bar(x + bar_width/2, s2.values, width=bar_width, label="2023", color="blue")

plt.xticks(x, month_order)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Comparison: 2022 vs 2023")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

```python
month_order= ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
gender_male = sf[sf['Gender']=="Male"]
gender_female= sf[sf['Gender']=="Female"]
g1 = gender_male.groupby("Month")["Price ($)"].sum()
g2=  gender_female.groupby("Month")["Price ($)"].sum()

g1.index = pd.CategoricalIndex(g1.index,categories=month_order,ordered=True)
g1=g1.sort_index()
g2.index = pd.CategoricalIndex(g2.index,categories=month_order,ordered=True)
g2=g2.sort_index()


x = np.arange(len(month_order))
bar_width = 0.45

plt.figure(figsize=(12,6))
plt.bar(x - bar_width/2, g1.values,width=bar_width, color="grey", label="Male")
plt.bar(x + bar_width/2, g2.values,width=bar_width, color="blue", label="Female")
plt.xticks(x,month_order)


plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales Comparison: Male vs female")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

```python
M = sf[sf['Dealer_Region']=="Austin"]                   
A = sf[sf['Dealer_Region']=="Janesville"]                           
G = sf[sf['Dealer_Region']=="Scottsdale"]                       

M1= M.groupby("Month")['Price ($)'].sum()                               
A1= A.groupby("Month")['Price ($)'].sum()                                   
G1= G.groupby("Month")['Price ($)'].sum()                                  

                                                     
month_order= ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
M1.index = pd.CategoricalIndex(M1.index,categories=month_order, ordered=True)
A1.index = pd.CategoricalIndex(A1.index,categories=month_order, ordered=True)
G1.index = pd.CategoricalIndex(G1.index,categories=month_order, ordered=True)

M1=M1.sort_index()
A1=A1.sort_index()
G1=G1.sort_index()

x = np.arange(len(month_order))
bar_width = 0.25

plt.figure(figsize=(11,5))
plt.bar(x - bar_width , M1.values, color='grey', width=bar_width, edgecolor="black",label="Austin")
plt.bar(x , A1.values, color='green', width=bar_width, edgecolor="black",label="Janesville")
plt.bar(x + bar_width, G1.values, color='blue', width=bar_width, edgecolor="black",label="Scottsdale")
plt.xticks(x,month_order)
plt.title("Monthly Sales Comparison: Selected Dealer Regions")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

```python
month_order= ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]

gender_male = sf[sf['Gender']=="Male"]
sale_by_gender_male = gender_male.groupby("Month")['Price ($)'].sum()
sale_by_gender_male.index=pd.CategoricalIndex(sale_by_gender_male.index,categories=month_order,ordered=True)
S1=sale_by_gender_male.sort_index()

gender_female = sf[sf['Gender']=="Female"]
sale_by_gender_female = gender_female.groupby("Month")['Price ($)'].sum()
sale_by_gender_female.index=pd.CategoricalIndex(sale_by_gender_female.index,categories=month_order,ordered=True)
S2= sale_by_gender_female.sort_index()


x= np.arange(len(month_order))
bar_width = 0.50

plt.figure(figsize=(12,6))
plt.bar(x,S1.values, width=bar_width, color='blue',label="Male" )
plt.bar(x,S2.values, bottom=S1.values,width=bar_width, color='grey',label="Female")
plt.xticks(x,month_order)


plt.xlabel("Month",color="blue")
plt.ylabel("Sales",color='blue')
plt.title("Sale By Month: Male Vs Female Analysis",color='blue')
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.hist(sf['Price ($)'], bins=20, color='skyblue', edgecolor='black')

plt.xlabel("Sales Amount ($)")
plt.ylabel("Frequency")
plt.title("Sales Distribution Histogram")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

```python
S = sf['Dealer_Region'].value_counts()

plt.figure(figsize=(12,8))
plt.pie(S,labels=S.index,autopct="%1.2f%%",explode=(0.1,0,0,0,0,0,0),shadow=True)
plt.title("Dealer Region Distribution",color="blue")
plt.legend()
plt.style.use('Solarize_Light2')
plt.savefig("dealer_region_distribution.png", bbox_inches="tight")
plt.show()
```
